## Cell 1 - Import library
NumPy untuk hitung, pandas untuk tabel iterasi, matplotlib untuk grafik.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Cell 2 - Definisi fungsi
Ubah isi fungsi untuk mencoba persamaan lain.

```python
def f(x):
    return x**3 - x - 2
```

## Cell 3 - Mengecek selang awal
Bisection hanya valid bila f(a) dan f(b) berbeda tanda (f(a)·f(b) < 0), sesuai Teorema Nilai Antara.

```python
a, b = 1.0, 2.0
print("f(a) =", f(a))
print("f(b) =", f(b))
print("Ada akar di [a,b]?", f(a) * f(b) < 0)
```

## Cell 4 - Fungsi Bisection
Setiap iterasi: titik tengah c = (a+b)/2, lalu pilih setengah selang yang tandanya masih berbeda. Berhenti jika setengah lebar selang < toleransi.

```python
def bisection(f, a, b, tol=1e-6, maks_iter=100):
    fa = f(a)
    if fa * f(b) > 0:
        raise ValueError("f(a) dan f(b) harus berbeda tanda")
    riwayat = []
    for i in range(1, maks_iter + 1):
        c = (a + b) / 2
        fc = f(c)
        galat = (b - a) / 2
        riwayat.append([i, a, b, c, fc, galat])
        if fc == 0 or galat < tol:
            break
        if fa * fc < 0:
            b = c
        else:
            a = c
            fa = fc
    tabel = pd.DataFrame(riwayat, columns=["iterasi", "a", "b", "c", "f(c)", "galat"])
    return c, tabel
```

## Cell 5 - Menjalankan dan menampilkan tabel

```python
akar, tabel = bisection(f, 1.0, 2.0, tol=1e-6)
pd.set_option("display.float_format", lambda v: f"{v:.7f}")
print(tabel.to_string(index=False))
print()
print(f"Akar hampiran : {akar:.7f}")
print(f"f(akar)       : {f(akar):.2e}")
print(f"Jumlah iterasi: {len(tabel)}")
```

## Cell 6 - Verifikasi
Banyak iterasi teoritis: n ≥ log2((b-a)/tol). Akar juga dibandingkan dengan solusi polinomial NumPy.

```python
n_teori = int(np.ceil(np.log2((2.0 - 1.0) / 1e-6)))
print("Iterasi teoritis :", n_teori)
akar_np = [r.real for r in np.roots([1, 0, -1, -2]) if abs(r.imag) < 1e-9][0]
print("Akar NumPy       :", round(akar_np, 7))
print("Selisih          :", abs(akar - akar_np))
```

## Cell 7 - Grafik
Kiri: kurva f(x) dan akar. Kanan: galat per iterasi (skala log), turun separuh tiap langkah.

```python
xs = np.linspace(0.8, 2.2, 200)
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(xs, f(xs), label="f(x)")
ax[0].axhline(0, color="gray", linewidth=0.8)
ax[0].scatter([akar], [0], color="red", zorder=3, label="akar")
ax[0].set_title("Kurva f(x) dan akar")
ax[0].legend()
ax[1].semilogy(tabel["iterasi"], tabel["galat"], marker="o")
ax[1].set_title("Penurunan galat tiap iterasi")
ax[1].set_xlabel("iterasi")
plt.tight_layout()
plt.show()
```
